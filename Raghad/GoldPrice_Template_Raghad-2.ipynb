## 1. Imports & Load Data

```python
# write your code here
import tensorflow as tf
from tensorflow import keras
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import VotingClassifier
from sklearn.neural_network import MLPClassifier

import matplotlib.pyplot as plt
```

```python
import kagglehub
import os

path = kagglehub.dataset_download("hayaalwizrah1/gold-price-prediction-dataset-20002026")
data = pd.read_csv(os.path.join(path, 'gold_data.csv'))
```

## 3. Define Features and Target

```python
# write your code here
y = df['Target']
X = df.drop(columns=["Date","Target", "Lag1","Gold_Close","Daily_Return"])
```

```python
y.shape, X.shape
```

## 4. Chronological Split (Train / Validation / Test)

```python
n = len(X)
train_end = int(n * 0.70)
val_end = int(n * 0.85)   # 0.70 + 0.15
```

```python
X_train = X[:train_end]
X_val = X[train_end:val_end]
X_test = X[val_end:]
```

```python
y_train = y[:train_end]
y_val = y[train_end:val_end]
y_test = y[val_end:]
```

```python
print(X_train.shape[0] == y_train.shape[0])
print(X_val.shape[0] == y_val.shape[0])
print(X_test.shape[0] == y_test.shape[0])
```

```python
X_train.shape[0] + X_val.shape[0] + X_test.shape[0]
```

## 5. Scaling

```python
# write your code here
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled.mean(axis=0)
```

## 6. Baseline

```python
# write your code here
baseline_pred = df['Gold_Close'][val_end:]      # today's price, used as the naive guess for tomorrow
baseline_mae = mean_absolute_error(y_test, baseline_pred)

print(baseline_mae)
```

```python
baseline_pred.shape[0] == y_test.shape[0]
```

```python
import os, random
import numpy as np
import tensorflow as tf

seed = 42
os.environ['PYTHONHASHSEED'] = str(seed)
random.seed(seed)
np.random.seed(seed)
tf.random.set_seed(seed)
```

## 7. Build & Compile & Train the Model (Dense Network)

```python
# write your code here
def build_model():
    model = keras.Sequential([
        keras.layers.Dense(256, activation='relu', input_shape=(20,)),
        keras.layers.Dropout(0.3),
        keras.layers.Dense(128, activation='relu'),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mae",
        metrics=["mae"]
    )
    return model

model = build_model()
```

```python
X_train_scaled.shape[1]
```

## 8. Early Stop

```python
# write your code here

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled, y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_val_scaled, y_val),
    callbacks=[early_stop]
)
```

## 9. Evaluation


```python
# write your code here
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.xlabel('Epoch')
plt.ylabel('Loss (MAE)')
plt.title('Training vs Validation Loss')
plt.legend()
plt.show()
```

```python
y_pred = model.predict(X_test_scaled).flatten()

test_mae = mean_absolute_error(y_test, y_pred)
test_mse = mean_squared_error(y_test, y_pred)
test_rmse = np.sqrt(mean_squared_error(y_test, y_pred))

test_r2 = r2_score(y_test, y_pred)

print("Test MAE:", test_mae)
print("Test MSE:", test_mse)
print("Test RMSE:", test_rmse)
print("Test R2:", test_r2)
```

```python
print("Baseline MAE:", baseline_mae)
print("Model MAE:", test_mae)
print("Improvement:", (baseline_mae - test_mae) / baseline_mae * 100, "%")
```
